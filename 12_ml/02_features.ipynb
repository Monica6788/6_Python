### 피처는 모델보다 중요하다
동일한 데이터로 알고리즘(모델)만 변경하여 얻는 성능 향상은\
보통 몇 퍼센트 정도가 최선인데 반해, **좋은 피처 하나를 추가하면**\
성능이 대폭 향상될 수 있다.

```python
from sklearn.ensemble import RandomForestClassifier
# 피처를 변경해가며 성능 비교
train, test, cutoff = time_split(df)

tr = train.sample(15_000, random_state=SEED)
te = test.sample(4_000, random_state=SEED)

조합 = {
    "1. 수익률만": ["ret_1d"],
    "2. +이동평균": ["ret_1d", "ma5_ratio", "ma20_ratio"],
    "3. 파생 변수 모두": FEATURES
}

for name, feats in 조합.items():
    # 모델
    m = RandomForestClassifier(n_estimators=50, max_depth=6,
                               random_state=SEED, n_jobs=-1)
    # 모델 학습시키기
    m.fit(tr[feats], tr["target_up"])
    print(f"{name:<20} 피처 {len(feats)}개 평가(정확도) {m.score(te[feats], te['target_up']):.4f}")
```

> **주가(시세) 예측은 원래 어렵다!\
> 피처를 늘리더라도 정확도가 크게 오르지는 않는다.\
> 50%에 근사하면 정상적인 결과라고 볼 수 있다.\
> **금융 시계열은 대부분 예측 불가한 잡음**이 존재하기 때문이다.

## 스케일링: 크기를 맞추는 작업

```python
df[FEATURES].agg(["mean", "std", "min", "max"]).T.round(4)
# .T : Pandas 데이터프레임의 Transpose(전치, 행과 열의 위치 바꾸기)
# 피처의 개수가 많아지면 옆으로 화면이 길어져 한 눈에 비교하기 불편함.
# row에 통계지표가 들어가고 열에 선택한 피처가 나열되도록 한 것임.
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

X_train, X_test = train[FEATURES], test[FEATURES]

scaler = StandardScaler()

# fit_transform 함수는 학습 데이터만 사용
#   fit (평균, 표준편차를 계산) + transform (계산된 기준으로 변환)
X_train_scaled = scaler.fit_transform(X_train)

# transform 함수로 테스트 데이터를 사용
X_test_scaled = scaler.transform(X_test)

print(f"학습 (변환 후): {X_train_scaled.mean():.6f} {X_train_scaled.std():.4f}")
print(f"테스트 (변환 후): {X_test_scaled.mean():.6f} {X_test_scaled.std():.4f}")
```

> **테스트 데이터의 평균이 정확하게 0이 아닌 것은 정상 결과임.**\
학습 데이터를 기준으로 평균, 표준편차로 변환했기 때문에,\
만약 테스트 데이터의 평균이 0, 표준편차가 1로 나온다면\
**테스트 정보로도 학습되었음**을 의심해야 함.

```python
# MinMaxScaler

mm = MinMaxScaler().fit(X_train)

mm_train = mm.transform(X_train)
mm_test = mm.transform(X_test)

print("=== 학습 데이터 범위 ===")
print(f"Standard Scaler: {X_train_scaled.min():.2f} ~ {X_train_scaled.max():.2f}")
print(f"MinMax Scaler: {mm_train.min():.2f} ~ {mm_train.max():.2f}")

print("=== 테스트 데이터 범위 ===")
print(f"Standard Scaler: {X_test_scaled.min():.2f} ~ {X_test_scaled.max():.2f}")
print(f"MinMax Scaler: {mm_test.min():.2f} ~ {mm_test.max():.2f}")
```

> **MinMax Scaler에서도 테스트 데이터의 범위가 [0, 1] 밖일 수 있다**\
학습 데이터의 최솟값, 최댓값 기준으로 계산되었으므로\
테스트 데이터에는 더 크거나 더 작은 값이 존재할 수 있다.

**트리 게열 모델인 경우**, 스케일링이 필요하지 않다.\
"거래량이 XX보다 큰가?"처럼 기준값으로 데이터를 나누는 방식이므로\
단위가 달라도 결과는 같다.

## 인코딩: 문자를 숫자로 변환
모델은 데이터를 숫자 형태로 받기 때문에,\
"XX전자"와 같은 문자열은 변환하여 전달해야 한다.

||LabelEncoder|One-Hot Encoding|
|:--:|:--:|:--:|
|결과|전기전자 -> 0, 금융 -> 1|범주마다 열 하나씩 (0/1)|
|열 개수|1개|범주 개수만큼 열이 늘어남|
|용도|순서가 있을 때|순서가 없을 때|

```python
# LabelEncoder
from sklearn.preprocessing import LabelEncoder

sample = pd.Series(["한식", "중식", "양식", "한식", "일식"])

le = LabelEncoder()
encoded = le.fit_transform(sample)

print(f"원본: {sample.tolist()}")
print(f"인코딩 후: {encoded.tolist()}")
```

```python
# One-Hot Encoding

# pandas 방식: pd.get_dummies()
onehot = pd.get_dummies(sample, prefix="food")
# 불리언 형식으로 나옴 => .astype(int)를 사용하여 숫자로 변환 가능
onehot_int = pd.get_dummies(sample, prefix="food").astype(int)

print(f"One-Hot 결과")
print(onehot.to_string(index=False))
print()

print(f"One-Hot (astype(int) 처리) 결과")
print(onehot_int.to_string(index=False))
```

**범주가 너무 많으면 One-Hot이 위험함**\
종목코드 120개를 One-Hot으로 처리하면...\
=> 열이 120개 늘어날 것임.\

데이터는 넓어지는데 각 열은 거의 0이 될 것임.\
=> 오히려 학습에 방해가 될 수 있음.\

> 범주가 많을 경우에는 몇 개만 남기고 나머지는 '기타'로 묶어 처리한다.

```python
top_n = 5

# .value_counts()로 종목코드별 행 수를 세고,
# top_n(5)개만큼 잘라서
# .index로 상위 5행의 코드만 추출
top_codes = full["code"].value_counts().head(top_n).index

# where(조건, 거짓일때바꿀값)을 사용하여 top_codes에 들어가면
# 해당 코드를 유지하고, 아니면 "기타"라는 코드로 처리
grouped = full["code"].where(full["code"].isin(top_codes), "기타")

print(f"기존 범주 수: {full['code'].nunique()}")
print(f"묶은 후: {grouped.nunique()}")
```

> **테스트 범주와 학습 범주가 다른 경우**\
학습 데이터에 없던 값이 테스트에 나타나면 열 구성이 달라져 오류가 발생됨.

```python
from sklearn.preprocessing import OneHotEncoder

학습_범주 = pd.DataFrame({"sector": ["금융", "화학", "금융"]})
테스트_범주 = pd.DataFrame({"sector": ["금융", "바이오"]})

enc = OneHotEncoder(sparse_output=False)
enc.fit(학습_범주)

try:
    enc.transform(테스트_범주)
except ValueError:
    print(f"기본 설정인 경우 ValueError 발생")
print()

enc2 = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
# handle_unknown="ignore": 학습(계산) 시 알 수 없는 값은 전부 0으로 처리
enc2.fit(학습_범주)
result = enc2.transform(테스트_범주)

print(f"열 이름: {enc2.get_feature_names_out().tolist()}")
print(f"결과: \n{result}")
# 테스트 데이터에 있는 "금융": 학습 때 알고 있던 것이므로 sector_금융 =1
#                   "바이오": 학습 때 본 적 없는 새로운 값이라서 0 처리
```

## 파생 변수 만들기
원본 열을 조합하여 **더 의미 있는 변수**를 만드는 것임.\
(도메인 지식이 반영되는 구간)

|원본|파생변수|의미|
|:--:|:--:|:--:|
|종가|n일 이동평균비율|평소보다 가격이 높은지, 낮은지|
|종가|전일대비수익률|절대가격보다 변화크기가 중요|
|거래량|20일 평균 대비 배수|거래량이 급증했는지?(터졌는지?)|

```python
two = full[full["code"].isin(["G0001", "G0002"])].copy()

# 20일 이동평균
two["ma20"] = (two.groupby("code")["close"]
                  .transform(lambda s: s.rolling(20).mean())
                )
# 20일 이동평균비율
two["ma20_ratio"] = two["close"] / two["ma20"]

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

# 코드 기준으로 그룹화 하여, 코드와 그룹화된 객체로 반복문
for code, g in two.groupby("code"):
    # plot(x축, y축, 라벨, lineweight)
    axes[0].plot(g["date"], g["close"], label=code, lw=1)
    axes[1].plot(g["date"], g["ma20_ratio"], label=code, lw=1)

axes[0].set_title("절댓값(종가)")
axes[0].set_xlabel("날짜")
axes[0].set_ylabel("종가(원)")

axes[1].set_title("비율(20일 평균대비)")
axes[1].set_xlabel("날짜")
axes[1].set_ylabel("종가(ma20)")

for ax in axes:
    ax.legend()
    ax.grid(alpha=0.3)

fig.tight_layout()
plt.show()

# 종가만 보면 급등 정도를 보기 어려운데,
# 비율을 보면 1.0을 기준으로 확 뛴 곳이 어딘지 알기 쉽다.
```

**대체로 절댓값보다는 비율이 더 파악하기 좋다.**\
왼쪽 그래프는 두 종목의 가격대가 다르다.\
=> 나란히 두고 비교하기 어렵다.\
오른쪽 그래프는 표시된 값이 1.0 근처에 차이를 보여주고 있다.\
=> 변화량에 대해 비교하기가 비교적 수월하다.

## 데이터 누수
**데이터 누수**: 학습 시점에 알 수 없는 정보가 피처에 섞여 들어가는 것

### 유형
- 미래정보: 오늘 종가로 오늘 피처를 만든 경우
- 정답누출: 정답에서 파생된 변수를 피처에 넣음
- 전처리 누수: 전체 데이터로 스케일링한 후에 분할
- 기타등등...

> **문제 재설정**\
> "당일 장이 종료되기 전에 오를지 예측"

당일의 종가는 아직 알 수 없음. 피처는 전일까지의 정보여야 함.
- 피처: 당일 종가 기준 이동평균 (shifit())\
=> 당일 종가는 끝나야 알 수 있기 때문이다.

```python
정상 = build_dataset(shift_features=True)
누수 = build_dataset(shift_features=False)

cols = ["date", "close", "ret_1d", "target_ret", "target_up"]

def display_data(data):
    """결과 출력용 데이터 반환"""
    v = data[data["code"] == "G0001"][cols].head(4).reset_index(drop=True)

    nums = ["ret_1d", "target_ret"]
    v[nums] = v[nums].round(5)

    return v

print("1. shift 적용 (정상)")
print(display_data(정상))
print()

print("2. shift 미적용 (누수)")
print(display_data(누수))
```

##### 누수 데이터를 확인하면
ret_1d와 target_ret의 값이 완전히 동일함.\
ret_1d(오늘의 수익률)와 target_ret(오늘의 수익률)이 같은 계산이므로\
정답을 피처로 준 것과 동일함.

```python
from sklearn.metrics import accuracy_score, roc_auc_score

결과 = []

for data, label in [(정상, "정상(shift_적용)"), (누수, "누수(shift_미적용)")]:
    tr_, te_, _ = time_split(data)

    # time_split을 사용하여 시계열을 기준으로 분할했으니 랜덤분할 ok
    tr_ = tr_.sample(15_000, random_state=SEED)
    te_ = te_.sample(4_000, random_state=SEED)

    m = RandomForestClassifier(n_estimators=50, max_depth=8, random_state=SEED, n_jobs=-1)

    m.fit(tr_[FEATURES], tr_["target_up"])           # 학습
    pred = m.predict(te_[FEATURES])                  # 예측 (결과)
    # 모든 행을 선택하고 인덱스 번호가 1인 열만 슬라이싱 하기
    # 0부터 숫자를 세므로 0번째 열이 하락(0)이고, 1열이 상승(1)일 것임.
    proba = m.predict_proba(te_[FEATURES])[:,1]      # 예측 (확률)
    # m.predict()는 1차원 배열, m.predict_proba()는 2차원 배열이라서 슬라이싱이 필요함.

    결과.append(
        {   
            # 위 for문에서 label에 담아둔 "정상(shift_적용)", "누수(shift_미적용)"
            "구분": label,
            "정확도": accuracy_score(te_["target_up"], pred),
            "AUC": roc_auc_score(te_["target_up"], proba)
        }
    )
print(pd.DataFrame(결과).round(4).to_string(index=False))
```

**정확도가 99% 이상 나올 수 있으나**\
실제로는 좋은 모델이 아님. 오늘의 종가를 미리 입력으로 줄 수 없음!\

**누수를 의심해야 하는 상황**\
- 성능이 기대보다 과하게 좋은 경우
- 특정 피처 하나의 중요도가 압도적으로 높을 경우
- 학습과 테스트의 성능 차이가 거의 없는 경우

누수 시에 피처를 전부 시각화 해보면 특정 피처와 정답이 동일하게 움직일 수 있음.\
-> 특정 피처의 중요도가 압도적인 경우(누수)

---
### 피처
: 모델에 영향을 주는 변수 (데이터)

### 스케일링
: 데이터 표준화 (크기를 맞추는 작업)\
 => fit_transform은 학습 데이터만 사용

### 인코딩
: 문자를 숫자로 변환 (순서가 없는 경우 LabelEncoder 사용 X)

### 파생 변수
: 기존 데이터로부터 만들어진 변수 (데이터). 절댓값보다는 비율로.

### 데이터 누수
: 학습 시점에 잘못된 데이터가 섞이는 현상. 성능이 너무 좋으면 의심할 것!

### 파이프라인 (Pipeline)
**순서/흐름**
1. 결측, 이상치 처리 (전처리)
2. 파생 변수 생성
3. 데이터 분리 (학습, 테스트)
4. 스케일링, 인코딩 -> fit(학습데이터)
5. 변환 -> transform(학습데이터), transform(테스트데이터)
...
...
...
