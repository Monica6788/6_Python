## 모델 저장과 활용

지금까지 만든 모델을 어떻게 남기고 재사용하는지 다루어보기.

코드 자체는 단 2줄!
```python
# 저장
joblib.dump(모델, "파일명")
# 불러오기
model = joblib.load("파일명")
```

````python
import os
import time

import joblib
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score

from _style import setup
from _ml import build_dataset, time_split, FEATURES, SEED

setup()
pd.set_option("display.width", 130)

df = build_dataset()
train, test, _ = time_split(df)

tr = train.sample(8_000, random_state=SEED)
te = test.sample(2_000, random_state=SEED)

X_train, y_train = tr[FEATURES], tr["target_up"]
X_test, y_test = te[FEATURES], te["target_up"]

print(f"학습 {len(tr):,}행 / 테스트 {len(te):,}행 / 피처 {len(FEATURES)}개")
````

||학습|예측|
|---|---|---|
|소요 시간|몇 분 ~ 몇 시간|1 ~ 2초|
|빈도|주 1회, 월 1회|**초당 수백 번**|
|필요한 것|전체 학습 데이터|모델 + 입력 하나|

````python
# 전처리 + 모델 (Pipeline)
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("model", RandomForestClassifier(n_estimators=100, 
                                     max_depth=8, 
                                     random_state=SEED, 
                                     n_jobs=-1))
])

# 학습 시간 측정
t0 = time.perf_counter()
pipe.fit(X_train, y_train)
train_s = time.perf_counter() - t0

# 예측 시간 측정 (1건)
single_row = X_test.head(1)

# 워밍업: 초기화 비용은 측정에서 제외

t1 = time.perf_counter()
for _ in range(50):
    pipe.predict(single_row)

# 여러 번 반복한 후 평균 계산
predict_s = (time.perf_counter() - t1) / 50

print(f"학습 1회: {train_s * 1_000:.1f}ms")
print(f"예측 1건: {predict_s * 1_000:.1f}ms")
print(f"차이 {train_s / predict_s:.0f}배")
print()
print(f"정확도: {pipe.score(X_test, y_test):.4f}")
````

> 💡**요청마다 학습하면 안 되는 이유**\
> 학습은 예측 1건보다 훨씬 오래 걸리고, 데이터가 커질수록 더 길어질 것이다.\
> **학습은 미리 해두고, 결과만 꺼내어 사용하는 것**이 저장하는 가장 큰 목적(이유)이다.

## joblib로 저장하고 불러오기

```python
joblib.dump(model, "model.pkl")     # 저장

```

두 줄이 끝!

````python
# model 파일을 저장하기 위한 폴더 생성 (models)
os.makedirs("models", exist_ok=True)

# ----- 저장 -----
joblib.dump(pipe, "models/pipeline.pkl")

size_kb = os.path.getsize("models/pipeline.pkl")
print(f"저장 완료: models/pipeline.pkl ({size_kb:,.0f}KB)")

# ----- 불러오기 -----
loaded = joblib.load("models/pipeline.pkl")

# 원래 모델과 같은 예측을 하는지 확인
original_pred = pipe.predict(X_test)
loaded_pred = loaded.predict(X_test)

print("불러오기 완료")
# array_equal(arr1, arr2): 두 배열이 같은지 확인하여 True/False 반환
print(f"예측 결과 비교: {np.array_equal(original_pred, loaded_pred)}")
````

> ⚠️ **모델은 서버가 구동될 때 한 번만 로드**
>
> 파일을 읽는 것도 비용이므로, 요청마다 `jobllib.load`를 하지 않고, 시작할 때 한 번 로드하여 메모리에 들고 있어야 한다.

> ⚠️ **출처를 모르는 pkl 파일은 열지 않는다.**
>
> `pickle` 형식은 **불러올 때 코드가 실행될 수 있는 구조**이므로, 내부적으로 만든 파일, 신뢰할 수 있는 출처의 파일만 사용하자.

## 스케일러 유/무 비교

`학습할 때 원본 데이터 + StandardScaler --> 모델`\
`예측할 때 새 데이터 ... --> 모델`\
스케일러가 없으면 새 데이터를 어떤 기준으로 변환할지 알 수 없다.

````python
# [1] 스케일러와 모델을 따로 두고, 모델만 저장
scaler = StandardScaler().fit(X_train)
model = RandomForestClassifier(n_estimators=100, max_depth=8,
                               random_state=SEED, n_jobs=-1)
# 저장하여 불러왔다고 가정하고 실습 진행

# 학습
model.fit(scaler.transform(X_train), y_train)

# + 스케일링 하여 예측
pred_scaled = model.predict(scaler.transform(X_test))
# + 스케일링 없이 예측
pred_unscaled = model.predict(X_test.values)

print(f"{'':<20}{'정확도':>10}")
print("-" * 30)
print(f"{'스케일링 후 예측':<19}{accuracy_score(y_test, pred_scaled):>10.4f}")
print(f"{'스케일링 없이 예측':<19}{accuracy_score(y_test, pred_unscaled):>10.4f}")
print()
print(f" 예측이 달라진 건수 {(pred_scaled != pred_unscaled).sum():,}"
      f" / {len(pred_scaled):,}"
      f" ({(pred_scaled != pred_unscaled).mean():.1%})")
````

> ⚠️ **오류가 발생하지 않음!**
> 
> 예측 결과가 **약 44% 달라졌지만** 아무런 경고도 없다. 정확도만 봐서는 오히려 조금 높게 확인된다. (물론 이는 우연이다.)
>
> **동작은 하지만 결과가 틀린 유형**의 문제라고 볼 수 있다. (데이터 누수)

## 피처 순서
모델의 경우 열 이름을 알 수 없고, **위치로만 판단**한다.\
거래량과 등락률의 순서를 바꾸면 어떻게 될까?


````python
import warnings

# 열 순서를 거꾸로 변경
reversed_X = X_test[FEATURES[::-1]]

print(f"원래 순서: {FEATURES[:3]} ...")
print(f"뒤집은 순서: {FEATURES[::-1][:3]} ...")

# [1] DataFrame 형태로 전달
try:
    loaded.predict(reversed_X)
    print(" * DataFrame 통과 * ")
except Exception as e:
    print(f" *  DataFrame: {type(e).__name__} ")
    print(str(e))

# [2] numpy 배열 형태로 전달
with warnings.catch_warnings():
    # warnings.simplefilter("ignore"): 경고만 끄고 실행
    warnings.simplefilter("ignore")
    pred_org = loaded.predict(X_test.values)
    pred_reversed = loaded.predict(reversed_X.values)

print(" * numpy 배열 *")
print(f"original: {len(pred_org):,} / reversed: {len(pred_reversed):,}")
print(f"예측이 달라진 건수: {(pred_org != pred_reversed).sum():,}건")
````

> **DataFrame** 형태로 전달하는 경우\
> 사이킷런에서 열 이름을 확인하고 순서가 잘못된 경우 예측을 막아준다. (예외를 발생시킨다.)

> **numpy 배열** 형태로 전달하는 경우\
> 사이킷런에서도 막아줄 수 없다. 배열에서 열 이름을 확인할 수 없으므로 순서를 확인할 방법이 없다.

**모델을 저장할 때 피처 목록도 함께 저장해야 한다.**\
예측 전에 저장된 순서대로 열을 정렬하여 진행할 수 있다.

## 모델 저장 시 함께 남겨야 하는 것
모델 파일 하나만 저장하면, 시간이 흐른 후 (몇 달 뒤) 기억할 수 있을까?

- **전처리기 (Pipeline)**: 동일한 변환을 적용해야 한다.
- **피처 목록과 순서**: 순서가 바뀌면 엉뚱한 예측을 할 수 있다.
- 학습 데이터 기간: 언제까지의 데이터로 학습을 시켰는지.
- 성능 지표: 나중에 새로 학습한 모델과의 비교 기준.
- 라이브러리 버전: 버전이 다를 경우 로드에 실패할 수 있다.

> 저장할 때 위의 정보들을 함께 pkl 형태로 저장할 수 있다.\
> (메타 데이터는 `meta.json`처럼 사람이 해석할 수 있는 파일로 따로 저장해두는 것이 좋음!)

````python
import sklearn

bundle = {
    # 전처리 + 모델
    "pipeline": pipe,
    # 피처 목록과 순서
    "features": FEATURES,
    # 학습 기간
    "train_period": (tr["date"].min().date(), tr["date"].max().date()),
    # 성능 지표
    "test_accuracy": pipe.score(X_test, y_test),
    # 라이브러리 버전
    "sklearn_version": sklearn.__version__,
    # 저장 날짜
    "created_at": time.strftime("%Y-%m-%d %H:%M:%S"),
}

joblib.dump(bundle, "models/model_bundle.pkl")
print("저장: models/model_bundle.pkl"
      f" ({os.path.getsize('models/model_bundle.pkl') / 1024:,.0f}) KB")
````

````python
# ... 수료 후 ...
b = joblib.load("models/model_bundle.pkl")

print("이 파일에 저장된 정보 (모델 정보)")
print(f"  학습 기간: {b['train_period'][0]} ~ {b['train_period'][1]}")
print(f"  테스트 정확도: {b['test_accuracy']:.4f}")
print(f"  라이브러리 버전: {b['sklearn_version']}"
      f" (현재 버전: {sklearn.__version__})")
print(f"  저장 날짜: {b['created_at']}")
print(f"  피처: {len(b['features'])} ({b['features'][:3]} ...)")
print()

pred = b['pipeline'].predict(X_test.head(5))
print(f"  예측 결과: ... {pred}")
````
