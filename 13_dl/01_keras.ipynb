````python
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 표준으로 사용되는 방식
# from tensorflow.keras import Sequential

# 공식 권장 방식
from keras import Sequential
from keras.layers import Dense, Dropout, Input
from keras.callbacks import EarlyStopping
# 한 가지 방식으로 통일하여 사용할 것!

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score, accuracy_score

import _ml
from _style import setup

from pprint import pprint

SEED = _ml.SEED
setup()

print("준비 완료")
````

---

머신러닝으로는 `오늘 오를까?`를 예측했고, 결과는 AUC 0.5 부근으로 사실상 랜덤이었다.\
신호가 없는 문제로는 **알고리즘이 잘 동작하는지조차 확인이 불가**했다.

딥러닝으로는 `오늘 거래량이 최근 20일 평균보다 많을까?`를 예측해본다. (거래량 급증 여부 예측)

거래량은 한 번 몰리면 며칠 이어지는 성질이 있어 **어느 정도는 예측이 가능**하다.

````python
def load_datasets():
    """
        거래량 급증 예측용 데이터셋을 만들어 반환

        - 기존 피처(_ml.FEATURES) 7개에 2개를 추가(volume_prev: 전일 거래량, close_prev: 전일 종가)
        - 정답 (surge): 오늘 거래량이 직전 20일 평균 거래량보다 크면 1, 아니면 0
    """
    # 기존 데이터셋에서 피처 2개 추가
    base = _ml.build_dataset().sort_values(["code", "date"])
    base["volume_prev"] = base.groupby("code")["volume"].shift(1)
    base["close_prev"] = base.groupby("code")["close"].shift(1)

    # 원본 데이터에서 계산 후 정답 추가
    raw = _ml.load_merged().sort_values(["code", "date"])

    # - 정답: 1 if 오늘 거래량 > 직전 20일 평균 거래량 else 0
    # 직전 20일 평균 거래량
    raw["vol_ma20_prev"] = (raw.groupby("code")["volume"]
                               .transform(lambda s: s.rolling(20)
                                                     .mean().shift()))
    # 정답 (surge): 거래량 급증 여부 (1: P 급증, 0: N)
    raw["surge"] = (raw["volume"] > raw["vol_ma20_prev"]).astype("int8")

    df = base.merge(raw[["code", "date", "surge"]], how="left", on=["code", "date"])
    return (df.dropna(subset=["volume_prev", "close_prev", "surge"])
              .reset_index(drop=True))

df = load_datasets()
FEATURES = _ml.FEATURES + ["volume_prev", "close_prev"]

print(f"데이터 {len(df):,}행, 피처 {len(FEATURES)}개")
print(f"급증 비용 {df['surge'].mean():,.3f}")
pprint(f"피처 {FEATURES}")
````

````python
# 시계열 데이터는 시간순으로 분할 (데이터 분리 -> 훈련용, 검증용)
train, test, cutoff = _ml.time_split(df, test_size=0.2)

# X(피처, 입력 데이터), y(정답, 레이블) 분리
X_train = train[FEATURES].to_numpy("float32")   # 넘파이 배열로 변경
y_train = train["surge"].to_numpy("float32")    # 데이터 타입 지정

X_test = test[FEATURES].to_numpy("float32")
y_test = test["surge"].to_numpy("float32")

print(f"분할 기준일: {cutoff.date()}")
print(f"학습: {X_train.shape} 검증: {X_test.shape}")
print()

scale = pd.DataFrame({
    "Mean": X_train.mean(axis=0),
    "Std": X_train.std(axis=0),
    "Max": X_train.max(axis=0),
}, index=FEATURES)

# to_string 함수 내부에도 포맷 지정 가능
print(scale.to_string(float_format=lambda v: f"{v:>18,.4f}"))
print()
print("Standard Deviation Max/Min = "
      f"{scale['Std'].max() / scale['Std'].min():,.0f}배 차이")
````

현재 표준편차가 **1억 배 이상 차이**를 보인다.\
이 상태로 신경망에 입력하면 거대한 피처가 기울기 계산을 독점하게 된다.\
-> 학습이 제대로 이루어지지 않으므로 **표준화 또는 정규화의 전처리가 필수적**이다.

### 딥러닝의 3단계: 모델 정의 -> 컴파일 -> 학습
1. 모델 정의: 어떤 층을 어떻게 쌓을 것인가
2. 컴파일: 무엇을 최소화하고 어떻게 학습할 것인가
3. 학습: 데이터를 넣고 반복

케라스(Keras)를 이용하면 통일된 형태로 간단히 구현이 가능하다.

````python
def build_model(n_features, hidden=(64, 32), dropout=0.0):
    """
        이진 분류용 신경망 생성
        
        Args
            n_features: 입력 피처 개수
            hidden    : 은닉층 뉴런 수. 튜플의 길이가 즉 은닉층 개수이다.
                        (기본 64, 32로 2층)
            dropout   : 0보다 큰 경우 각 은닉층 뒤에 Dropout을 추가 (비율, 0~1)

        Return
            model     : 컴파일까지 끝난 Sequential 모델
    """
    # 가중치 초기값 고정. (동일한 데이터로 사용하도록 설정)
    # 동일한 실험 결과를 얻기 위해 난수 시드 설정
    tf.keras.utils.set_random_seed(SEED)

    # 입력층 (입력 벡터의 크기 -> 피처 개수)
    # 신경망 층들을 담을 리스트 초기화 및 입력층 추가
    # 입력 데이터의 형태를 피처 개수 크기의 1차원 벡터로 지정
    layers = [Input(shape=(n_features, ))]

    for units in hidden:
        layers.append(
            # 은닉층 (활성화 함수: 계산이 단순, 비선형 표현)
            # 은닉층의 units에 대해 Dense객체 생성.
            # 지정된 뉴런(units) 수와 ReLU 활성화 함수를 가진 밀집층 추가
            Dense(units, activation="relu")
        )
        if dropout > 0:
            # dropout > 0이면 Dropout 객체의 rate를 dropout으로 하여 append
            # 과적합 방지용
            layers.append(Dropout(dropout))
    # 이진 분류 -> 출력 1개 + sigmoid
    layers.append(Dense(1, activation="sigmoid"))

    # [1] 모델 정의
    #     앞서 쌓은 레이어 리스트(layers)를 일렬로 줄 세워 하나의 인공신경망 뼈대로 조립
    model = Sequential(layers)

    # [2] 컴파일
    #     조립된 뼈대에 달릴 엔진(Optimizer)과 오차를 잴 눈금(Loss) 세팅
    model.compile(
        # 파라미터별 학습 속도를 적응적으로 조절하는 최적화 방법(알고리즘)
        optimizer="adam",
        # 손실 함수 -> 오차를 측정하는 기준 
        # 이진 분류에서는 흔히 아담과 바이너리크로스엔트로피를 짝꿍처럼 같이 사용한다.
        loss="binary_crossentropy",
        # 정확도(accuracy) 기록 설정
        metrics=["accuracy"]
    )

    return model

model = build_model(len(FEATURES))
# model.summary()
# : 생성된 신경망의 층 구조, 출력 형태, 학습해야 할 가중치 개수 출력
model.summary()
````

**Total params**: Dense 층의 파라미터 = (입력개수 * 뉴런개수) + 뉴런개수(편향)\
dense_n     = 9 x 64 + 64 = 640\
dense_(n+1) = 64 x 32 + 32 = 2080\
dense_(n+2) = 32 x 1 + 1 = 33\
Total = 2753

(실행할 때마다 (dense, dense_1, dense_2), (dense_3, dense_4, dense_5), ...와 같이 이름이 바뀌어서 위와 같이 표현함.)\
1번 층 - 입력개수 = 9: 피처 개수\
         뉴런개수: 64: 기본값으로 준 hidden의 64\
2번 층 - 입력개수 = 64: 이전 층의 결과물로 나온 데이터의 개수. 1번 층에서 뉴런 개수를 64개로 만들어서 계산을 끝낸 후 64개의 값을 다음 층으로 넘긴다.\
         뉴런개수 = 32: 기본값으로 준 hidden의 32\
3번 층 - 입력개수 = 32: 2번 층에서 넘어온 데이터의 개수\
         뉴런개수 = 1: 이진 분류 모델이므로 최종 결과는 확률 1개. 따라서 Dense(1, activation="sigmoid")로 출력 뉴런을 1개만 생성했다.

Dense(64) 한 줄로 가중치 576개와 편향 64개가 자동 생성된다.\
Dropout 층은 파라미터가 없고, 파라미터 수는 데이터 행 수와 무관하게 모델 구조로만 정해진다.

### 학습 곡선을 그리는 함수 정의

````python
def plot_history(hist, title, ax=None, ylim=None):
    """
        학습 곡선을 그리는 함수

        Args
            hist : model.fit()이 반환한 History 객체
            title: 차트 제목
            ax   : 그래프를 그릴 영역 객체 (Axes)
                   None인 경우 새로 생성
            ylim : y축 범위 (min, max)
        
        Return
            ax   : 학습 곡선이 그려진 그래프 객체 (Axes)
    """
    if ax is None:
        _, ax = plt.subplots(figsize=(8, 4))
    if ylim:
        ax.set_ylim(*ylim)

    # loss: 훈련 데이터 오차
    # val_loss: 검증 데이터 오차
    epochs = range(1, len(hist.history['loss']) + 1)

    ax.plot(epochs, hist.history['loss'], label="train", lw=1.6)
    ax.plot(epochs, hist.history['val_loss'], label="validation", lw=1.6)

    # 검증 손실이 가장 낮았던 지점
    # argmin(): 가장 낮은 값의 인덱스
    best= int(np.argmin(hist.history['val_loss'])) + 1

    ax.axvline(best, color="gray", ls=":", lw=1)
    ax.set(xlabel="Epoch", ylabel="Loss", title=title)
    ax.legend()

    return ax

def evaluate(name, model, X, y, scaler=None):
    """
        주어진 데이터(X, y)에 대한 모델 성능을 한 줄로 출력

        Args
            name   : 출력할 모델 이름
            model  : 학습된 모델
            X      : 입력 피처 (scaler 사용 시 스케일 전 원본 값을 전달해야 함)
            y      : 정답 타깃
            scaler : 학습 데이터로 fit된 스케일러 (None이면 X를 그대로 사용)

        Return
            auc: ROC AUC (float)
        
        Print
            AUC와 정확도 - 출력 예시: {name} AUC {auc} 정확도 {acc}
    """
    # 스케일러가 있으면 변환, 없으면 원본 데이터를 사용하여 Xe 준비
    Xe = X if scaler is None else scaler.transform(X)

    # verbose=0: 예측 과정을 진행할 때 콘솔창에 로딩 바나 로그가 뜨지 않도록
    #            조용히 숨겨서 실행하는 옵션
    # .ravel(): 2차원 배열 형태로 나올 예측 결과를 1차원 배열로 flatten
    prob = model.predict(Xe, verbose=0).ravel()

    auc = roc_auc_score(y, prob)
    # 기준값 = 0.5로 정확도
    acc = accuracy_score(y, (prob >= 0.5).astype(int))

    print(f"{name:<28} AUC {auc:.4f} ACC {acc:.4f}")
    return auc
````

### 스케일링 없이 학습
원본 피처를 그대로 학습 (거래량은 수천만, 등략률 0.0x)

````python
# 검증 구간 만들기
# 1) 시간순 정렬 (과거 -> 최신 순)
#    kind="stable": 날짜가 같을 때 원래 순서를 유지해주기 위한 옵션
order = np.argsort(train['date'].to_numpy(), kind="stable")

# 전체 데이터 개수의 20%를 검증용 데이터로 쓰기 위해 변수로 저장
n_val = int(len(order) * 0.2)
# 과거순 정렬된 order 데이터를 20% 포인트로 슬라이싱
# order[:-n_val]: 뒤에서 n_val개 빼고 나머지 전부
# order[-n_val:]: 뒤에서부터 n_val개
fit_idx, val_idx = order[:-n_val], order[-n_val:]

X_fit, y_fit = X_train[fit_idx], y_train[fit_idx]
X_val, y_val = X_train[val_idx], y_train[val_idx]

print(f"학습 {X_fit.shape} 검증 {X_val.shape}")
print(f"검증구간 시작일: {train['date'].iloc[val_idx].min()}")
````

````python
def print_history(hist, digits=4):
    """
        학습 기록을 최대 10줄로 줄여 출력하는 함수

        Args
            hist   : fit() 후 전달받은 History 객체
            digits : 출력할 때 단위 옵션 (ex. .2f, .3f)
    """
    loss, val_loss = hist.history['loss'], hist.history['val_loss']

    step = max(1, len(loss) // 10)
    print("epoch        train_loss    val_loss")
    for i in range(0, len(loss), step):
        # epoch는 0부터 시작이니까 +1
        print(f"{i + 1:>5} {loss[i]:>14,.{digits}f} {val_loss[i]:>12,.{digits}f}")
````

````python
t0 = time.time()

model_raw = build_model(len(FEATURES))
hist_raw = model_raw.fit(
    X_fit, y_fit, epochs=20,
    batch_size=512, validation_data=(X_val, y_val),
    shuffle=False, verbose=0,    #출력 옵션 -  출력 끄기
)

print(f"학습 시간: {time.time() - t0:.0f}초")
print_history(hist_raw, digits=1)
````

#### 확인된 패턴
- 손실이 처음부터 **수백 ~ 수천** 단위. (정상적이라면 0.6 ~ 0.7 근처에서 시작되어야 함.)
- train/val 손실이 **증감을 반복**하며 안정적으로 내려가지 않고 있다.

--> 과적합보다는 **학습 불안정(발산)**에 해당한다.\
    (과적합은 train은 계속 감소하는데 val만 증가하는 상태)

````python
plot_history(hist_raw, "스케일링 없이 학습된 모델")
plt.show()

auc_raw = evaluate("스케일링 없이 학습된 모델", model_raw, X_test, y_test)
print(f"AUC {auc_raw:.2f}")
````

### 스케일링 후 학습
> StandardScaler로 모든 피처를 평균 0, 표준편차 1로 맞추기

````python
scaler = StandardScaler().fit(X_fit)

X_fit_s = scaler.transform(X_fit)
X_val_s = scaler.transform(X_val)
X_test_s = scaler.transform(X_test)

print(
    pd.DataFrame({"Mean": X_fit_s.mean(axis=0).round(3),
                  "Std": X_fit_s.std(axis=0).round(3)}, index=FEATURES)
)
````

````python
t0 = time.time()

model_scaled = build_model(len(FEATURES))
hist_scaled = model_scaled.fit(
    X_fit_s, y_fit,
    epochs=20, batch_size=512,
    validation_data=(X_val_s, y_val),
    shuffle=False, verbose=0
)

print(f"학습 시간: {time.time() - t0:.0f}초")
print_history(hist_scaled)
````

````python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

plot_history(hist_raw, "스케일링 없음 (손실단위: 수천)", ax=axes[0])
plot_history(hist_scaled, "스케일링 후 (손실단위: 0.6x)", ax=axes[1])

plt.tight_layout()
plt.show()

auc_raw = evaluate("스케일링 없음", model_raw, X_test, y_test)
auc_scaled = evaluate("스케일링 후", model_scaled, X_test_s, y_test)

print(f"AUC 비교 (스케일링 x -> o) : {auc_raw:.3f} -> {auc_scaled:.3f}")
````

> 스케일링은 단순히 성능을 조금 올려주는 것이 아닌, 신경망 자체의 동작 여부 문제로 볼 수 있다.
>
> 그러므로 **스케일러와 모델은 항상 함께 다루어야** 한다.

#### 히스토리 해석
|모양|상태|할 일|
|---|---|---|
|둘 다 감소|학습이 되는 중|에포크를 늘린다|
|둘 다 낮고 붙어 있음|학습 완료 상태|그대로 사용한다|
|train이 감소할 때 val은 증가|과적합|학습을 중단시킨다|
|둘 다 높고 감소하지 않음|과소적합, 학습데이터/피처 부족|스케일링을 확인하고 -> 모델을 키우고 -> 피처를 추가한다|
|둘 다 크게 튀며 불안정|학습 불안정(발산)|스케일링을 확인하고 -> 학습률을 낮춘다|

**모델을 키운다?**
모델의 용량(Capacity)를 늘려준다는 뜻으로, 구체적으로는 다음 2가지 방법 중 하나(또는 둘 다)를 의미한다.\
1. 층(Layer)을 더 깊게: 은닉층을 추가하여 모델이 더 복잡하고 고차원적인 패턴을 학습하도록 만든다.
2. 뉴런(Units)을 더 많이: 앞서 보았던 hidden(64, 32) 설정에서 숫자를 키운다. 

**왜 모델을 키울까?**
학습 데이터의 손실도 높고 검증 데이터의 손실도 높으면서 줄어들지 않는 상태(과소적합)는 "모델의 뇌가 너무 단순해서 복잡한 데이터를 감당하지 못하는 상태"이다. 따라서 모델의 덩치(파라미터 수)를 키워주어야 데이터를 제대로 학습할 수 있습니다.

- 이렇게 **두 곡선이 벌어지는 모양이 "과적합"이다.**

### Dropout + EarlyStopping 적용
- Dropout(0.3): 학습 중 뉴런의 30%를 무작위로 끔 -> 특정 뉴런에 의존하지 못하게 한다.
- EarlyStopping: patience값만큼 참고, 이 값이 넘어가도 검증 손실이 나지 않으면 멈춘다.

TODO

````python
es = EarlyStopping(
    # 무엇을 볼 것인가: 검증 데이터 손실값 val_loss 감시
    monitor="val_loss",
    # 몇 번 참을 것인가: 성능이 개선되지 않아도 8번의 에포크까지는 기다리기
    patience=8,
    # 참기 전 가장 좋았던 시점으로 되돌리기
    # : 학습 중단 시 가장 성능이 좋았던 때의 가중치(weights)로 모델 복원
    restore_best_weights=True
)

t0 = time.time()

# 지정된 피처 개수, 은닉층 구조, 드롭아웃 (0.3)을 이용하여 신경망 모델 생성
model_drop = build_model(len(FEATURES), hidden=(256, 256, 128), dropout=0.3)
print(f"{model_drop.count_params():,}개 / 학습 데이터 {len(X_fit):,}행")

# 모델 학습 시작, 학습한 모델은 hist_drop 변수에 저장
hist_drop = model_drop.fit(
    X_fit_s, y_fit,
    # 최대 60번 반복, 한 번에 512개씩 묶어 학습
    epochs=60, batch_size=512,
    # 검증용 데이터셋을 주고 에포크마다 성능 평가
    validation_data=(X_val_s, y_val),
    # 데이터를 섞지 않고, 학습 결과 출력도 생략(verbose=0)
    # 학습 중 조기 종료(es) 콜백을 적용하여 조건 만족 시 자동 중단
    # callbacks=[es]: 학습을 진행하는 동안 es 규칙을 지켜보다가, 조건 충족 시 강제 학습 중단
    # es: 앞에서 정의한 EarlyStopping 객체
    shuffle=False, verbose=0, callbacks=[es]
)

# .history["loss"]: 딕셔너리 형태의 데이터 속성
# 모델이 학습하는 동안 매 에포크마다 기록된 학습 성적표.
# 매 학습 에포크마다 loss나 accuracy가 어떻게 변화하는지 기록하고 딕셔너리 형태로 담아둔 것.
ran = len(hist_drop.history["loss"])
best = int(np.argmin(hist_drop.history['val_loss'])) + 1

print(f"실행한 에포크: {ran} val_loss 최저 에포크: {best}")
print(f"학습 시간: {time.time() - t0:.0f}초")
print_history(hist_drop)
````

### 모델 저장
Keras에서는 모델 저장 시 `joblib`이 아닌 `model.save()`를 사용한다.

````python
import joblib
from keras.models import load_model

from _config import model_path, MODELS_DIR

# keras 모델 저장
# model_drop: 위 코드셀에서 학습을 마친 모델
model_drop.save(model_path("surge_nn.keras"))

# 스케일러를 별도로 저장
joblib.dump(scaler, model_path("surge_scaler.pkl"))
# 피처 정보 저장
joblib.dump(FEATURES, model_path("surge_features.pkl"))

for f in sorted(os.listdir(MODELS_DIR)):
    size = os.path.getsize(os.path.join(MODELS_DIR, f))
    print(f"{f} {size:.1f} KB")
````

````python
# 다시 불러와서 사용하기
loaded_model = load_model(model_path("surge_nn.keras"))
loaded_scaler = joblib.load(model_path("surge_scaler.pkl"))
loaded_features = joblib.load(model_path("surge_features.pkl"))

# DataFrame을 dtype="float32"인 Numpy 배열로 변환
raw_input = test[loaded_features].to_numpy("float32")[:5]

# 저장 전 모델로 예측 (.ravel()을 이용하여 1차원 배열로 펼쳐주기)
# (데이터수, 1) 형태의 DataFrame으로 나오기 때문에 출력을 편하게 보려고 1차원으로
before = model_drop.predict(scaler.transform(raw_input), verbose=0).ravel()

# 저장된 모델로 예측
after = loaded_model.predict(loaded_scaler.transform(raw_input), verbose=0).ravel()

print(f"저장 전 불러온 결과: {before.round(5)}")
print(f"불러온 후 예측 결과: {after.round(5)}")

# np.allclose(): 미세한 오차는 같다고 치기 위해 == 대신 allclose 사용
# atol=1e-6: Absolute Tolerance(절대 오차 허용치)를 10의 -6승으로 지정
print(f"일치 여부: {np.allclose(before, after, atol=1e-6)}")
````

모델 저장 시 함께 저장해야 하는 것: 모델 + 스케일러 + 피처 정보(목록/순서)

- 모델만 저장하면 불러온 후 입력을 똑같이 스케일링 할 수 없고, 피처 순서도 알 수 없다.
- `joblib.load`는 신뢰할 수 있는 파일에만 사용해야 한다.
